# Annotations

```python
batters = pd.read_csv('datasets-session-24/batter.csv')
batters.head()
```

```python
sample_df = batters.head(100).sample(25, random_state=5)
sample_df
```

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.scatter(x,y)
plt.text(1,5,'Point 1')
plt.text(2,6,'Point 2')
plt.text(3,7,'Point 3')
plt.text(4,8,'Point 4',fontdict={'size':12,'color':'brown'})
```

# Horizontal and Vertical Lines

```python
plt.figure(figsize=(18,10))

plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130, color = 'red') # Horizontal line
plt.axvline(30, color= 'red') # vertical Line

plt.axhline(140, color = 'green') # Horizontal line
plt.axvline(35, color= 'green') # vertical Line

for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
# batter dataset

fig,ax = plt.subplots(figsize = (15,10), nrows = 2, ncols = 1, sharex= True)

ax[0].scatter(batters['avg'],batters['strike_rate'], color= 'red')
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')

ax[1].set_xlabel('Avg')
```

```python
# batter dataset

fig,ax = plt.subplots(figsize = (10,6), nrows = 2, ncols = 2)

ax[0,0].scatter(batters['avg'],batters['strike_rate'], color= 'red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])

ax[0,0].set_title('Avg Vs Strike Rate')
ax[0,0].set_ylabel('Strike Rate')

ax[0,1].set_title('Avg Vs Runs')
ax[0,1].set_ylabel('Runs')

```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'], color= 'red')

ax2 = fig.add_subplot(2,2,2)
ax2.scatter(batters['avg'],batters['runs'], color= 'green')

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['runs'])
```

# 3D Surface Plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
xx,yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
```

```python
fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

# Contour Plots

```python
# it represents 3d graph -> 2d graph

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
# it represents 3d graph -> 2d graph

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z, cmap = 'viridis')
fig.colorbar(p)
```
