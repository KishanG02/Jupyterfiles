Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool built on top of python programming language.

Pandas series is like a column in a table. It is a 1-D array holding data of any type.

## Importing Pandas

```python
import numpy as np
import pandas as pd
```

## Series from a list

```python
# string

country = ['India', 'USA', 'China', 'Japan', 'Germany']

pd.Series(country)
```

```python
# integers
runs = [13,24,56,78,100]

runs_series = pd.Series(runs)
```

```python
# custom index

marks = [76, 88, 75, 97]
subjects = ['Maths', 'Physics', 'Chemistry', 'Biology']

pd.Series(marks, index=subjects)
```

```python
# setting a name

pd.Series(marks, index=subjects, name='Marks Obtained')
```

# Series from Dictionary

```python
marks = {
    'Maths' : 76,
    'Physics' : 88,
    'Chemistry' : 75,
    'Biology' : 97
}

marks_series = pd.Series(marks, name='Marks Obtained')
type(marks_series)
```

```python
# sort_index -> inplace -> movies
# Output the series sorted by its index (row labels)
# uses the sort_index() method to sort the series by its index (row labels) in ascending order by default, which can be useful for organizing data or preparing it for further analysis.

movies.sort_index()
```
