```python
# write a function that can return the track record of 2 teams against each other

def compare(team1, team2):
    mask1 = ((ipl['Team1'] == team1) & (ipl['Team2'] == team2))
    mask2 = ((ipl['Team1'] == team2) & (ipl['Team2'] == team1))

    total_match = ipl[(mask1 | mask2)].shape[0]

    win_team1 = ipl[(mask1 | mask2) & (ipl['WinningTeam'] == team1)].shape[0]
    win_team2 = ipl[(mask1 | mask2) & (ipl['WinningTeam'] == team2)].shape[0]

    perc_team1 = round((win_team1/total_match)*100,2)
    perc_team2 = round((win_team2/total_match)*100,2)

    return total_match,win_team1,win_team2,perc_team1,perc_team2


print(compare('Chennai Super Kings','Sunrisers Hyderabad'))

```

# apply (series + dataframe)

```python
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
def sigmoid(value):
    return 1/1+np.exp(-value)
```

```python
temp = pd.Series([10,20,30,40,50])
temp.apply(sigmoid)
```

```python
def euclidean(row):
    pt_a = row['1st point']
    pt_b = row['2nd point']

    return ((pt_a[0] - pt_b[0])**2 + (pt_a[1] - pt_b[1])**2)**0.5
```

```python
points_df['distance'] = points_df.apply(euclidean,axis=1)
points_df
```

```python
# Looping on groups

for group,data in genres:
    print(data[data['IMDB_Rating']== data['IMDB_Rating'].max()])
```

```python
def foo(group):
    return group['Series_Title'].str.startswith('A').sum()
    
```

```python
# split (apply) combine
# apply -> builtin function

genres.apply(foo)
```

```python
def rank_movie(group):
    group['genre_rank'] = group['IMDB_Rating'].rank(ascending = False)
    return group
```

```python
# find ranking of each movie in the group according to IMDB score.

genres.apply(rank_movie)
```

```python
def normalized(group):
    group['Normalized_Rating'] = (group['IMDB_Rating'] - group['IMDB_Rating'].min())/(group['IMDB_Rating'].max() - group['IMDB_Rating'].min())
    return group
```

```python
# find normalized IMDB rating group wise
genres.apply(normalized)
```
