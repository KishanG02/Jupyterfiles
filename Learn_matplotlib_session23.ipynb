```python
# import the library

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# 2D Line plot

- used for Bivariate Analysis
- categorical -> numerical and numerical -> numerical
- use case Time series data

```python
# plotting a simple function

price = [48000,54000,57000,49000,47000,45000]
year = [2015,2016,2017,2018,2019,2020]

# x-axis -> categorical & y-axis -> Numerical

plt.plot(year,price)
```

```python
# from a pandas dataframe
batsman = pd.read_csv('datasets-session-23/sharma-kohli.csv')
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
# plotting multiple plots
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
# Colour (hex) and line(Width and style) and marker (Size)

plt.plot(batsman['index'],batsman['V Kohli'],color = 'red')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red', linestyle = 'solid')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green', linestyle = 'dashdot', linewidth = 3)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color = 'red', linestyle = 'solid', marker = 'D')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green', linestyle = 'dashdot', linewidth = 3, marker = 'o', markersize = 7)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')
```

```python
# legend -> location

plt.plot(batsman['index'],batsman['V Kohli'],color = 'red', linestyle = 'solid', marker = 'D', label = 'Virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green', linestyle = 'dashdot', linewidth = 3, marker = 'o', markersize = 7, label = 'Rohit')

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')

# loc = best
plt.legend(loc = 'upper right')
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,450000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)

plt.ylim(0,75000)
plt.xlim(2017,2019)

# generally we use it when there is a outlier.
```

```python
# grid

plt.plot(batsman['index'],batsman['V Kohli'],color = 'red', linestyle = 'solid', marker = 'D')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green', linestyle = 'dashdot', linewidth = 3, marker = 'o', markersize = 7)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')

plt.grid()
```

```python
# show

plt.plot(batsman['index'],batsman['V Kohli'],color = 'red', linestyle = 'solid', marker = 'D')
plt.plot(batsman['index'],batsman['RG Sharma'],color = 'green', linestyle = 'dashdot', linewidth = 3, marker = 'o', markersize = 7)

plt.title('Rohit Sharma Vs Virat Kohli Career Comparison')

plt.xlabel('Season')
plt.ylabel('Runs')

plt.grid()
# used if we are not using proper console.
plt.show()
```

# Scatter Plots

- Bivariate Analysis
- Numerical vs Numerical
- Use case - Finding Correlation

```python
# plt.scatter simple function

x = np.linspace(-10,10,50)
y = 10*x + 3 + np.random.randint(0,300,50)
```

```python
plt.scatter(x,y)
```

```python
# plt.scatter on pandas data
df = pd.read_csv('datasets-session-23/batter.csv')
df = df.head(50)

plt.scatter(df['avg'],df['strike_rate'],color = 'red', marker='+')
plt.title('Avg and SR analysis of Top 50 Batsman')
plt.xlabel('Avg')
plt.ylabel('Strike_Rate')
plt.grid()
```

```python
# size
tips = sns.load_dataset('tips')
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

```python
# scatterplot using plt.plot
# this is a faster technique
# but some functionality cannot work in this.

plt.plot(tips['total_bill'],tips['tip'],'o')
```

# Bar chart

- Bivariates Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [20,15,40,30,35]
colors = ['red','green','blue','yellow','pink']

plt.bar(colors, children ,color = ['red','green','blue','yellow','pink'])
```

```python
# horizontal bar chart
plt.barh(colors, children ,color = ['red','green','blue','yellow','pink'])
```

```python
# color and label

df = pd.read_csv('datasets-session-23/batsman_season_record.csv')

plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
```

```python
# Multiple bar chart
plt.bar(np.arange(df.shape[0])-0.2,df['2015'], width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'], width=0.2)
plt.bar(np.arange(df.shape[0])+0.2,df['2017'], width=0.2)
# xticks
plt.xticks(np.arange(df.shape[0]),df['batsman'])
```

```python
# a problem
children = [20,15,40,30,35]
colors = ['red red red red red','green green green green','blue blue blue blue blue','yellow yellow yellow yellow ','pink pink pink pink']

plt.bar(colors, children ,color = ['red','green','blue','yellow','pink'])
plt.xticks(rotation = 'vertical')
```

```python
# Stacked bar chart

df
```

```python
plt.bar(df['batsman'],df['2015'],label = '2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label = '2016')
plt.bar(df['batsman'],df['2017'], bottom=df['2016']+df['2015'], label='2017')

plt.legend()
```
